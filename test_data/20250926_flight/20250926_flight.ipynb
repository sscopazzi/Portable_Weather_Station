```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import inset_axes

# --- Load data ---
df = pd.read_csv("20250926_flight.csv", header=0)
df = df[800:2100]

# --- Main plot ---
fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(df.index, df["pressurehPa"], label="Pressure", lw=3)

# --- Labels and formatting ---
ax.set_title("Pressure (hPa) During Takeoff", fontsize=24)
ax.set_xlabel(f"Sample Number (no GNSS time in plane, approx {len(df)/60:.1f}min)", fontsize=18)
ax.set_ylabel("Pressure (hPa)", fontsize=18)
ax.set_ylim(820, 1026)
ax.grid(alpha=0.5)
# ax.legend(loc="lower left", markerscale=10)
ax.tick_params(axis="both", labelsize=14)

# --- Inset plot (zoom on 900–1000) ---
axins = inset_axes(ax, width="40%", height="40%", loc="upper right", borderpad=2)
axins.plot(df.index, df["pressurehPa"], color="tab:blue", lw=3)
axins.set_xlim(850, 1000)
axins.set_ylim(1013.5, 1019.5)
axins.grid(alpha=0.5)
axins.tick_params(axis="both", labelsize=12)  # inset ticks

# --- Vertical lines ---
ax.vlines(x=922, ymin=0, ymax=2000, colors="red", linestyles="--", lw=3, label="Takeoff")
axins.vlines(x=922, ymin=0, ymax=2000, colors="red", linestyles="--", lw=3)

plt.show()
```
